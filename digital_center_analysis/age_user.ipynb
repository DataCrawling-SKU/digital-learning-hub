```python
import pandas as pd
```

```python
age_df = pd.read_csv('C:/Users/wjdah/vscode-workspace/digital-learning-hub/digital_center_analysis/dataFile/age_learning.csv', sep=',', header=0, engine='python')
```

```python
age_df.describe()
```

```python
# 연도별 배움터 교육인원 합계 구하기
year_df_sum = age_df.groupby('연도')['디지털 배움터 교육인원(명)'].sum().reset_index() # reset_index를 써서 데이터프레임으로 만듦
year_df_sum
```

```python
# 연령별 배움터 교육인원 합계 구하기
age_df_sum = age_df.groupby('연령')['디지털 배움터 교육인원(명)'].sum().reset_index() # reset_index를 써서 데이터프레임으로 만듦
age_df_sum
```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as tk

plt.rc('font', family='Malgun Gothic')
plt.rcParams['axes.unicode_minus'] = False

plt.figure(figsize=(12, 5))

year_df_sum.index = ['2020', '2021', '2022', '2023', '2024']
colors = ['skyblue', 'skyblue', 'skyblue', 'skyblue', 'red']
plt.bar(year_df_sum['연도'], year_df_sum['디지털 배움터 교육인원(명)'], color = colors)

plt.title('연도별 배움터 교육인원')
plt.xlabel('연도')
plt.ylabel('교육인원')

plt.gca().ticklabel_format(style='plain', axis='y')
plt.gca().yaxis.set_major_formatter(tk.StrMethodFormatter('{x:,.0f}'))

plt.xticks([2020, 2021, 2022, 2023, 2024])

plt.show()
```
